## Generate .env Template File

```python
# Generate comprehensive .env template
def generate_env_template():
    lines = ["# ═══════════════════════════════════════════════════════════════"]
    lines.append("# BUILDSPACES ENVIRONMENT CONFIGURATION")
    lines.append(f"# Generated: {datetime.now().isoformat()}")
    lines.append("# ═══════════════════════════════════════════════════════════════")
    lines.append("")
    
    for category, config_dict in all_configs.items():
        lines.append(f"\n# {'='*60}")
        lines.append(f"# {category.upper()}")
        lines.append(f"# {'='*60}")
        
        for key, config in config_dict.items():
            lines.append("")
            lines.append(f"# {config['description']}")
            if config.get('hint'):
                lines.append(f"# Hint: {config['hint']}")
            if config.get('example'):
                lines.append(f"# Example: {config['example']}")
            
            if 'default' in config:
                lines.append(f"{key}={config['default']}")
            else:
                lines.append(f"# {key}=")
    
    return "\n".join(lines)

env_template = generate_env_template()
print("✓ Environment template generated!")
print(f"\nTemplate Preview (first 50 lines):")
print("="*70)
for i, line in enumerate(env_template.split('\n')[:50]):
    print(line)
```

## GitHub Actions Secrets Format

```python
# Generate GitHub Actions secrets format
print("\n📋 GITHUB ACTIONS SECRETS FORMAT")
print("=" * 70)
print("\nTo add these to GitHub Actions, go to:")
print("Repository Settings → Secrets and variables → Actions → New repository secret")
print("\nSecrets (sensitive variables that should be masked):\n")

secrets_list = []
for category, config_dict in all_configs.items():
    for key, config in config_dict.items():
        if config.get('sensitive', False):
            secrets_list.append(key)

for key in sorted(secrets_list):
    print(f"  • {key}")

print(f"\n\nTotal Secrets: {len(secrets_list)}")

print("\n" + "="*70)
print("Non-sensitive variables (can be added as repository variables):\n")

vars_list = []
for category, config_dict in all_configs.items():
    for key, config in config_dict.items():
        if not config.get('sensitive', False):
            vars_list.append(key)

for key in sorted(vars_list):
    default = " [has default]" if "default" in [v.get('default') for v in all_configs.values()] else ""
    print(f"  • {key}{default}")

print(f"\nTotal Variables: {len(vars_list)}")
```
